```python

import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get("OpenAi_api")
os.environ["TAVILY_API_KEY"] = userdata.get("TAVILY_API_KEY")

```

```python
search_tool=TavilySearchResults(max_results=2)
```

```python
result=search_tool.invoke("what is the capital of france")
result
```

```python
llm=ChatOpenAI(
    model="gpt-3.5-turbo",
    temperature=0,
    max_tokens=500,
    OPENAI_API_KEY=os.environ["OpenAi_api"]
)
```

```python
prompt=hub.pull("hwchase17/react")
```

```python
tools=[search_tool]
```

```python
agent= create_agent(
    llm=llm,
    tools=tools,
    prompt=prompt,
    verbose=True
)
```

```python
response=agent.invoke()
```

Configure your GitHub credentials below. It is highly recommended to store your GitHub Personal Access Token (PAT) securely in Colab's **Secrets** (the key icon 🔑 on the left panel) under the name `GITHUB_TOKEN`.
